Sampling Distribution

````python
sample_mean=[]
````

````python
for i in range(1000):

  sample=np.random.choice(
      nilai,
      size=2,
      replace=False
  )

  mean=np.mean(sample)

  sample_mean.append(mean)
````

````python
populasi=np.random.normal(
  loc=70, #mean
  scale=5, #std deviasi
  size=10000 #jumlah sampel
)
````

````python
plt.hist(populasi,bins=30)
plt.title("Histogram Populasi")
plt.xlabel("Tinggi Badan")
plt.ylabel("Jumlah Pelanggan")
plt.show()
````

Scipy

````python
from scipy import stats

distribusi_normal = stats.norm.rvs(
    loc=70, #mean
  scale=5, #std deviasi
  size=10000 #jumlah sampel
)

plt.hist(distribusi_normal,bins=30)
plt.title("Distribusi Normal Populasi")
plt.xlabel("Tinggi Badan")
plt.ylabel("Jumlah Pelanggan")
plt.show()
````

````python
from scipy import stats

distribusi_exponential = stats.expon.rvs(
    scale=5, #std deviasi
    size=10000 #jumlah sampel
)

plt.hist(distribusi_exponential,bins=30)
plt.title("Distirbusi Exponential Populasi")
plt.xlabel("Tinggi Badan")
plt.ylabel("Jumlah Pelanggan")
plt.show()
````

````python
from scipy import stats

distribusi_poisson = stats.poisson.rvs(
    mu=5, #rata-rata
    size=10000 #jumlah sampel
)

plt.hist(distribusi_poisson,bins=30)
plt.title("Distribusi Poisson Populasi")
plt.xlabel("Tinggi Badan")
plt.ylabel("Jumlah Pelanggan")
plt.show()
````
